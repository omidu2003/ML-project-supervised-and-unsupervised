```python
import cv2
from fer import FER
from google.colab.patches import cv2_imshow

def detect_emotions_in_image(image_path, mtcnn=False):
    # Load the image
    image = cv2.imread(image_path)

    # Initialize the FER detector
    if mtcnn == True:
        detector = FER(mtcnn=True)
        print("Using mtcnn network for face detection, slower but more accurate")
    else:
        detector = FER()
        print("Using Haar cascade classifier for face detection, slower but more accurate")

    # Detect emotions
    emotions = detector.detect_emotions(image)

    # Draw bounding boxes and label emotions
    if emotions:
        for face in emotions:
            # Retrieve bounding box coordinates
            (x, y, w, h) = face['box']

            # Draw bounding box
            cv2.rectangle(image, (x, y), (x+w, y+h), (0, 255, 0), 2)

            # Label top emotion
            top_emotion = max(face['emotions'], key=face['emotions'].get)
            cv2.putText(image, top_emotion, (x+5, y+h-10), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2, cv2.LINE_AA)

    # Display the image
    cv2_imshow(image)

def detect_emotions_in_video(video_path):
    # Initialize video capture
    cap = cv2.VideoCapture(video_path)

    # Initialize the FER detector
    detector = FER()

    prev_emotion = None

    # Define the desired output size
    output_width = 108*3
    output_height = 192*3

    # Get the total number of frames in the video
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    # Get the frame rate of the video
    frame_rate = cap.get(cv2.CAP_PROP_FPS)

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        # Resize the frame
        frame = cv2.resize(frame, (output_width, output_height))

        # Detect emotions
        emotions = detector.detect_emotions(frame)

        # Check if there's a change in emotion
        new_emotion = None
        if emotions:
            new_emotion = max(emotions[0]['emotions'], key=emotions[0]['emotions'].get)

        if new_emotion != prev_emotion:
            # Display the emotions detected
            if emotions:
                # Get the predominant emotion
                predominant_emotion = max(emotions[0]['emotions'], key=emotions[0]['emotions'].get)
                (x, y, w, h) = emotions[0]['box']
                cv2.putText(frame, f"{predominant_emotion}", (x + 5, y + h - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2, cv2.LINE_AA)
                cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)

            # Get the current frame number
            frame_number = int(cap.get(cv2.CAP_PROP_POS_FRAMES))

            # Get the current time in the video
            current_time = frame_number / frame_rate

            # Print frame information
            print(f"Frame {frame_number}/{total_frames}, Time: {current_time:.2f} seconds")

            # Display the frame
            cv2_imshow(frame)
            prev_emotion = new_emotion

        # Exit if 'q' is pressed
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    # Release video capture
    cap.release()
    cv2.destroyAllWindows()

# Example usage for image
#image_path = '/content/disgusting.jpg'
#detect_emotions_in_image(image_path, True)
#detect_emotions_in_image(image_path)

# Example usage for video
video_path = '/content/drive/MyDrive/WhatsApp Video 2025-03-03 at 10.14.27_f2652798.mp4'
detect_emotions_in_video(video_path)
```
